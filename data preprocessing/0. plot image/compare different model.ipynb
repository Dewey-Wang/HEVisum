```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

def summarize_scores(data):
    # 如果是 dict，先轉成 DataFrame
    if isinstance(data, dict):
        df = pd.DataFrame(data)
    else:
        df = data.copy()

    df_long = df.melt(
        id_vars="submission",
        var_name="model",
        value_name="score"
    )

    summary = (
        df_long
        .groupby("model")["score"]
        .agg(["mean", "std"])
        .reset_index()
    )
    return summary

summary_public  = summarize_scores(df_public)
summary_private = summarize_scores(df_private)
summary_local_CV = summarize_scores(df_local_CV)

```

```python
# 依 private mean 由高到低排序
order = (
    summary_private
    .sort_values("mean", ascending=False)["model"]
    .values
)
summary_private_ord = (
    summary_private
    .set_index("model")
    .loc[order]
    .reset_index()
)
summary_local_CV_ord = (
    summary_local_CV
    .set_index("model")
    .loc[order]
    .reset_index()
)

summary_public_ord = (
    summary_public
    .set_index("model")
    .loc[order]
    .reset_index()
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

models = summary_private_ord["model"]
x = np.arange(len(models))
width = 0.38

plt.figure(figsize=(12, 6))

width = 0.25

plt.figure(figsize=(12, 6))

plt.bar(
    x - width,
    summary_public_ord["mean"],
    width,
    yerr=summary_public_ord["std"],
    capsize=4,
    label="Public (25% data)",
    color="#4C72B0"
)


plt.bar(
    x,
    summary_local_CV_ord["mean"],
    width,
    yerr=summary_local_CV_ord["std"],
    capsize=4,
    label="Local CV",
    color="#CBDD52"
)

plt.xticks(x, models)
plt.ylabel("Spearman Correlation")
plt.title("Pipeline Comparision (Ordered by Private Score)")
plt.legend(
    loc="upper left",
    bbox_to_anchor=(1.02, 1)
)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# 路徑
path_a = "output_folder/try_embed_pca_har/submission_fold_mean.csv"
path_b = "output_folder/try_r2/submission_rank_ensemble.csv"
out_path = "output_folder/submission_ensemble.csv"

# 讀檔
df_a = pd.read_csv(path_a)
df_b = pd.read_csv(path_b)

# ===== 用 ID 對齊（避免順序錯）=====
df = df_a.merge(df_b, on="ID", suffixes=("_a", "_b"))

# ===== 對每個 cell type 取平均 =====
cell_cols = [f"C{i+1}" for i in range(35)]

for c in cell_cols:
    df[c] = 0.5 * df[f"{c}_a"] + 0.5 * df[f"{c}_b"]

# 只留下 submission 需要的欄位
df_out = df[["ID"] + cell_cols]

# 存檔
df_out.to_csv(out_path, index=False)
print(f"✅ Saved ensemble submission to {out_path}")
```
